## Theory & Dataset Information

### Theory
Pandas and Matplotlib are two fundamental, open-source Python libraries that work in tandem for data analysis and visualization:
* **Pandas:** A powerful tool for data manipulation and analysis, providing high-performance, easy-to-use data structures like the DataFrame (a 2D table) and Series (a 1D labelled array).
* **Matplotlib:** A versatile plotting library that serves as the underlying engine for creating a wide variety of static, animated, and interactive visualizations in Python.

### Dataset Details
* **Dataset:** Classic Iris Dataset (Fisher, 1936).
* **Instances:** 150 plants (50 instances for each of the 3 species: *setosa*, *versicolor*, *virginica*).
* **Predicted Attribute:** Class / Species of Iris plant.
* **Missing Values:** No missing values.
* **Source:** UCI Machine Learning Repository (https://archive.ics.uci.edu/dataset/53/iris)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('iris_dataset.csv')

# Display the first few rows to inspect column names
df.head()
```

---
## Matplotlib Plots

### 1. Single Line Plot
**Task:** Plot 'SepalLengthCm' (or `sepal length (cm)`) using a red dashed line (`'r--'`).

```python
plt.figure(figsize=(9, 4))
plt.plot(df['sepal length (cm)'], 'r--', label='Sepal Length (cm)')
plt.title('Single Line Plot: Sepal Length')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### 2. Multiline Plot
**Task:** Drop non-numeric columns like `target` and `species` and plot all numerical columns as separate lines.

```python
# Drop non-numeric columns for the multiline plot
numeric_df = df.drop(columns=['target', 'species'], errors='ignore')

plt.figure(figsize=(10, 5))
plt.plot(numeric_df)
plt.title('Multiline Plot of Iris Numerical Features')
plt.xlabel('Sample Index')
plt.ylabel('Feature Value (cm)')
plt.legend(numeric_df.columns, loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### 3. Scatter Plot
**Task:** Generate a scatter plot of Sepal Length vs. Petal Length, differentiating the species by color.

```python
plt.figure(figsize=(8, 6))

species_list = df['species'].unique()
colors = ['red', 'green', 'blue']

for species, color in zip(species_list, colors):
    subset = df[df['species'] == species]
    plt.scatter(
        subset['sepal length (cm)'], 
        subset['petal length (cm)'], 
        label=species, 
        color=color, 
        alpha=0.7, 
        edgecolors='k'
    )

plt.title('Scatter Plot: Sepal Length vs. Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend(title='Species')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### 4. Histogram of a Single Feature
**Task:** Plot a Histogram of Sepal Width.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['sepal width (cm)'], bins=15, color='skyblue', edgecolor='black', alpha=0.8)
plt.title('Histogram of Sepal Width')
plt.xlabel('Sepal Width (cm)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

### 5. Box Plot of All Features
**Task:** Plot Box and Whisker for each numerical attribute.

```python
numeric_df = df.drop(columns=['target', 'species'], errors='ignore')

plt.figure(figsize=(9, 5))
numeric_df.plot(kind='box', grid=True, patch_artist=True, boxprops=dict(facecolor='lightyellow', color='black'))
plt.title('Box and Whisker Plot for Each Numerical Attribute')
plt.ylabel('Values (cm)')
plt.xlabel('Features')
plt.show()
```
