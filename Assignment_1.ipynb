## Theory

Statistical analysis helps us organize, summarize, and interpret data. Statistical measures describe the central tendency and dispersion of a dataset.

### 1. Measures of Central Tendency

**Mean:** The mean is the sum of all observations divided by the total number of observations. It represents the average value of the dataset.

**Formula:** Mean = Sum of observations / Number of observations

**Median:** The median is the middle value of a dataset after arranging the observations in ascending order.
- If the number of observations is odd, the median is the middle value.
- If the number of observations is even, the median is the average of the two middle values.

### 2. Measures of Dispersion

**Variance:** Variance measures how far observations are spread out from the mean. It is calculated by taking the average of the squared differences between each observation and the mean.

**Formula:** Population variance = Sum of squared differences from the mean / Number of observations

**Standard Deviation:** Standard deviation is the square root of the variance. It indicates the spread of observations around the mean and is expressed in the same unit as the original data.

These measures help us understand the average, distribution, and variability of a dataset.

## Algorithm

1. Start the program.
2. Read the number of observations.
3. If the number of observations is not positive, display an error message.
4. Read the data values.
5. Calculate the mean by adding all values and dividing by the number of observations.
6. Sort the data in ascending order using the bubble sort technique.
7. Calculate the median depending on whether the number of observations is odd or even.
8. Calculate the population variance by summing the squared differences from the mean and dividing by the number of observations.
9. Calculate the standard deviation using an iterative square-root method.
10. Display the sorted data, mean, median, variance, and standard deviation.
11. Stop the program.

## Python Program

The following program uses loops and basic arithmetic. It does not import any library or use built-in statistical functions such as `mean()`, `median()`, `variance()`, or `stdev()`.


````python
# Program to calculate Mean, Median, Variance and Standard Deviation
# Without external libraries or built-in statistical functions

data = []

n = int(input("Enter the number of observations: "))

if n <= 0:
    print("Please enter a positive number of observations.")
else:
    print("Enter the observations:")

    for i in range(n):
        value = float(input())
        data.append(value)

    # Calculate Mean
    total = 0
    for i in range(n):
        total = total + data[i]

    mean = total / n

    # Sort data using Bubble Sort
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if data[j] > data[j + 1]:
                temp = data[j]
                data[j] = data[j + 1]
                data[j + 1] = temp

    # Calculate Median
    if n % 2 == 1:
        median = data[n // 2]
    else:
        median = (data[n // 2 - 1] + data[n // 2]) / 2

    # Calculate Population Variance
    squared_difference = 0
    for i in range(n):
        difference = data[i] - mean
        squared_difference = squared_difference + difference * difference

    variance = squared_difference / n

    # Calculate Standard Deviation using Newton's method
    if variance == 0:
        standard_deviation = 0
    else:
        standard_deviation = variance
        for i in range(100):
            standard_deviation = (
                standard_deviation + variance / standard_deviation
            ) / 2

    # Display Results
    print("\\nSorted Data:", data)
    print("Mean:", mean)
    print("Median:", median)
    print("Variance:", variance)
    print("Standard Deviation:", standard_deviation)
````
